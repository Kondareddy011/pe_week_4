# Experiment 4: Diagnosing Prompt Failures & Edge Cases

Regenerated as a standalone notebook using Google Gemini API.

```python

import os, json, re, getpass
try:
    from google import genai
except ImportError as exc:
    raise ImportError("Install the Gemini SDK first: %pip install -q google-genai") from exc

# Use a new Gemini API key. Input is hidden and not saved in this notebook.
API_KEY = os.getenv("GEMINI_API_KEY") or getpass.getpass("Enter your Gemini API key (input hidden): ")
MODEL = "gemini-2.5-flash"
client = genai.Client(api_key=API_KEY)

def ask_llm(prompt, system="You are a helpful assistant.", temperature=0.2):
    response = client.models.generate_content(
        model=MODEL,
        contents=prompt,
        config={"system_instruction": system, "temperature": temperature}
    )
    return response.text or ""

print("Gemini client configured. Model:", MODEL)
```

## Experiment Program

```python
cases = [
 ("Hallucination", "Who won the FIFA World Cup in 1942?", "The FIFA World Cup was not held in 1942 because of World War II. If known, identify the previous tournament winner."),
 ("Format", "List the three largest oceans on Earth.", "List the three largest oceans using exactly three dash-prefixed bullet points."),
 ("Benign request", "Tell me a fun fact about the Roman Empire appropriate for school children.", "Give one school-appropriate Roman Empire fact in one sentence.")
]
for label, original, corrected in cases:
    print(f"\\n=== {label} ===")
    print("Original:", ask_llm(original))
    print("Corrected:", ask_llm(corrected))
```
